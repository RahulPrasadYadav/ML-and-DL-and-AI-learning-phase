# Map() Function 

```python
def square(x):
    return x*x
```

```python
number=[1,2,3,3,44,5,6,7,7]

list(map(square,number))
```

```python
## lmbda fucntion on map 


number=[1,2,3,3,44,5,6,7,7]

list(map(lambda x: x*x, number))
```

```python
#muliply two list using map


num1=[1,2,3,4,5]
num2=[6,7,8,9,10]

added_num=list(map(lambda x, y: x + y, num1, num2))

print(added_num)
```

```python
word=['heelo','ArithmeticError','hello','world']


upper_word=list(map(str.upper,word))
print(upper_word)
```
